```python
import pandas as pd

# Read data from two sheets in Online Retail II
path = '../data/online_retail_II.xlsx'
df_sheet1 = pd.read_excel(path, sheet_name='Year 2009-2010')
df_sheet2 = pd.read_excel(path, sheet_name='Year 2010-2011')
```

```python
df_merged = pd.concat([df_sheet1, df_sheet2], ignore_index=True)
```

```python
df_merged.to_csv('../data/retail_II_2009_2011.csv', index=False)
```
